## Create and inspect the batch control table

This notebook creates and seeds `company_db.batch_control.batch_table`, which is used to track batch processing state.

It performs the following steps:

* Creates or replaces the batch control table
* Optionally inserts sample batch records for testing
* Queries the table to verify the contents
* Keeps a commented cleanup statement that can be used to reset the table during development

This notebook is useful for initial setup and for validating the control-table structure used by the downstream orchestration notebooks.

```python
catalog = dbutils.widgets.get("catalog")
container = dbutils.widgets.get("container")
storage_account = dbutils.widgets.get("storage_account")
```

```python
%run "../setup/002_Environment_variables"
```

```python

# Create or replace the control table used to track each batch.
# The table stores the batch identifier, its current processing status,
# and audit timestamps for when the record was created and last updated.
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {catalog}.batch_control.batch_table
(
  batch_id string,
  batch_status STRING,
  created_at TIMESTAMP,
  updated_at TIMESTAMP
)
USING DELTA
LOCATION 'abfss://{container}@{storage_account}.dfs.core.windows.net/batch_control/batch_table';""")
```
