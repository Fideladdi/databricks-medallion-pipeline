## Create new batch control record

This notebook creates a new entry in `company_db.batch_control.batch_table` for the batch ID provided at runtime.

It performs the following steps:

* Reads the `id` input parameter from a notebook widget
* Inserts a new row into the batch control table
* Marks the batch status as `in_progress`
* Stamps the record with the current timestamp for create and update times
* Prints a confirmation message for orchestration logs

This notebook is typically used after a previous task has identified the next batch to process and passed that batch ID into this notebook.

```python
catalog = dbutils.widgets.get("catalog")
container = dbutils.widgets.get("container")
storage_account = dbutils.widgets.get("storage_account")
```

```python
%run "../setup/002_Environment_variables"
```

```python
# Import Spark SQL helper functions for any dataframe transformations used in this notebook.
from pyspark.sql import functions as F
```

```python
# Create a text widget so the notebook can receive the batch ID at runtime.
# In job runs, this value is typically passed from an upstream task.
batch_id = dbutils.widgets.text('id',"")

# Read the widget value into a Python variable for use in the insert statement.
batch_id = dbutils.widgets.get('id')

# Insert a new record into the batch control table to register this batch as in progress.
# The current timestamp is recorded for both the created and updated audit fields.
spark.sql(
f"""
    insert into {catalog}.batch_control.batch_table
    values
    ('{batch_id}', 'in_progress', current_timestamp(), current_timestamp())

""")

# Print a confirmation message so the run output clearly shows which batch was registered.
print(f'Batch_id: {batch_id} in progress')
```
