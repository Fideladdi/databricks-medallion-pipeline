#### Step 1: Create the external location
This cell registers the Azure Data Lake Storage path as a Unity Catalog external location.

What this step is intended to do:
* Connect Databricks to the ADLS container used for the project
* Associate the storage path with the configured storage credential
* Provide a governed storage location for downstream catalog and volume setup


#### Step 2: Create the catalog
This cell creates the main Unity Catalog catalog that will contain the project schemas.

What this step is intended to do:
* Create a dedicated catalog for Company data assets
* Set the managed storage location for catalog-managed objects
* Keep project data organized under one top-level namespace


#### Step 3: Create the schemas
This cell creates the schemas used for the medallion architecture and landing zone.

What this step is intended to do:
* Separate data into `bronze`, `silver`, `gold`, and `landing` layers
* Assign managed storage locations to the curated schemas
* Prepare the database structure for ingestion and transformation notebooks


#### Step 4: Create the external volume
This cell creates an external volume for raw files in the landing layer.

What this step is intended to do:
* Expose the landing storage path as a Unity Catalog volume
* Provide a governed file location for source CSV and raw uploads
* Make raw files accessible to ingestion notebooks through a stable path


```python
%sql
-- Register the ADLS container as a Unity Catalog external location.
-- This links the cloud storage path to the storage credential so it can be governed and reused.
CREATE EXTERNAL LOCATION IF NOT EXISTS company_external_storage_adls
URL 'abfss://company-dev@stcompanydev.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `company_storage_credential_dev`)
COMMENT 'Creating external location to landing';
```

```python
%sql
-- Create the main Unity Catalog catalog for the Company project.
-- The managed location defines where catalog-managed data will be stored in ADLS.
CREATE CATALOG IF NOT EXISTS company_dev
  MANAGED LOCATION 'abfss://company-dev@stcompanydev.dfs.core.windows.net';
```

```python
%sql
-- Create the main Unity Catalog catalog for the Company project.
-- The managed location defines where catalog-managed data will be stored in ADLS.
CREATE CATALOG IF NOT EXISTS company_test
  MANAGED LOCATION 'abfss://company-test@stcompanytest.dfs.core.windows.net';
```

```python
%sql
-- Create the main Unity Catalog catalog for the Company project.
-- The managed location defines where catalog-managed data will be stored in ADLS.
CREATE CATALOG IF NOT EXISTS company_prod
  MANAGED LOCATION 'abfss://company-prod@stcompanyprod.dfs.core.windows.net';
```

```python
%sql
-- Create schemas for each layer of the medallion architecture.
-- Bronze, silver, and gold use dedicated managed locations, while landing is used for raw file access and setup.
CREATE SCHEMA IF NOT EXISTS company_dev.bronze
    MANAGED LOCATION 'abfss://company-dev@stcompanydev.dfs.core.windows.net/bronze';

CREATE SCHEMA IF NOT EXISTS company_dev.silver
    MANAGED LOCATION 'abfss://company-dev@stcompanydev.dfs.core.windows.net/silver';

CREATE SCHEMA IF NOT EXISTS company_dev.gold
    MANAGED LOCATION 'abfss://company-dev@stcompanydev.dfs.core.windows.net/gold';

CREATE SCHEMA IF NOT EXISTS company_dev.batch_control
    MANAGED LOCATION 'abfss://company-dev@stcompanydev.dfs.core.windows.net/batch_control';

CREATE SCHEMA IF NOT EXISTS company_dev.landing;
```
